## 2. Değişkenler ve Tipler
Bir değişken, bir değere bağlanan isimdir. Python tipi otomatik olarak belirler.

```python
musteri_adi = "Ayşe Demir"
siparis_tutari = 249.90
vip_mi = True

print(musteri_adi, siparis_tutari, vip_mi)
print(type(musteri_adi), type(siparis_tutari), type(vip_mi))
```

## 3. Operatörler
Aritmetik: `+ - * / // % **`  |  Karşılaştırma: `== != > < >= <=`  |  Mantıksal: `and or not`

```python
adet = 12
birim_fiyat = 45.0
indirim_orani = 0.10 if adet >= 10 else 0.0

ara_toplam = adet * birim_fiyat
toplam = ara_toplam * (1 - indirim_orani)
print("Ara toplam:", ara_toplam, "| İndirim sonrası toplam:", toplam)
```

## 4. Koşullar — `if` / `elif` / `else`
**İş senaryosu:** üyelik seviyesi ve sipariş büyüklüğüne göre sadakat indirimine karar verin.

```python
def indirim_orani(seviye, siparis_tutari):
    if seviye == "gold":
        return 0.15
    elif seviye == "silver" and siparis_tutari > 500:
        return 0.10
    else:
        return 0.0

print(indirim_orani("silver", 620))
print(indirim_orani("bronze", 620))
```

## 5. Döngüler — `for` ve `while`
**İş senaryosu:** bir grup siparişi toplayın ve risk eşiğinin üzerindeki siparişleri işaretleyin.

```python
# for: bilinen bir koleksiyonu dolaş
siparisler = [120.50, 980.00, 45.25, 1200.00, 300.75]
birikmis_toplam = 0

for tutar in siparisler:
    birikmis_toplam += tutar
    isaret = " <-- YÜKSEK DEĞER, incele" if tutar > 1000 else ""
    print(f"Sipariş: {tutar:>8.2f} | Birikmiş toplam: {birikmis_toplam:>9.2f}{isaret}")
```

```python
# while: bir koşul False olana kadar tekrarla
deneme = 0
while deneme < 3:
    print("Giriş denemesi", deneme + 1)
    deneme += 1
```

## 7. Fonksiyonlar
Mantığı yeniden kullanılabilir, isimlendirilmiş bir bloğa paketleyin. Parametreler girdidir; `return` bir sonucu geri gönderir.

```python
def fatura_toplami(birim_fiyat, adet, vergi_orani=0.18):
    ara_toplam = birim_fiyat * adet
    vergi = ara_toplam * vergi_orani
    return ara_toplam + vergi

print(fatura_toplami(150, 3))
print(fatura_toplami(150, 3, vergi_orani=0.08))
```

### Çözümler (önce kendiniz deneyin, sonra karşılaştırın)

```python
siparisler = [430.0, 1250.0, 75.5, 999.99, 60.0]
birikmis_toplam = 0
for tutar in siparisler:
    birikmis_toplam += tutar
    isaret = " <-- YÜKSEK DEĞER" if tutar > 1000 else ""
    print(f"{tutar:>8.2f} | birikmiş toplam: {birikmis_toplam:>9.2f}{isaret}")

def indirim_uygula(tutar, seviye):
    if seviye == "gold":
        return tutar * 0.85
    elif seviye == "silver":
        return tutar * 0.90
    else:
        return tutar

print(indirim_uygula(1000, "gold"))
```
