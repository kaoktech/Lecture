## 2. İş Senaryosu: Kredi Ön Onay Algoritması
İyi bir algoritmanın beş özelliğini izleyin: net girdi, net çıktı, kesin adımlar, sonlanma ve her adımın yürütülebilir olması.

````python
def kredi_on_onay(gelir, mevcut_borc, kredi_skoru):
    borc_gelir_orani = mevcut_borc / gelir
    if kredi_skoru >= 700 and borc_gelir_orani < 0.35:
        return "onaylandı"
    elif kredi_skoru >= 600 and borc_gelir_orani < 0.45:
        return "manuel inceleme gerekiyor"
    else:
        return "reddedildi"

print(kredi_on_onay(60000, 15000, 720))
print(kredi_on_onay(50000, 20000, 650))
print(kredi_on_onay(30000, 20000, 550))
````

## 3. Ayrıştırma: Bir Siparişi İşlemek
"Bu siparişi işle"yi doğrula → hesapla → karar ver şeklinde, her biri kendi fonksiyonu olacak biçimde bölün.

````python
def siparisi_dogrula(sepet, envanter):
    return all(envanter.get(urun, 0) > 0 for urun in sepet)

def toplam_hesapla(sepet, fiyatlar, vergi_orani=0.18):
    ara_toplam = sum(fiyatlar[urun] for urun in sepet)
    return ara_toplam * (1 + vergi_orani)

def siparisi_isle(sepet, envanter, fiyatlar):
    if not siparisi_dogrula(sepet, envanter):
        return "Sipariş reddedildi: ürün stokta yok"
    return round(toplam_hesapla(sepet, fiyatlar), 2)

envanter = {"laptop": 5, "fare": 0, "monitor": 12}
fiyatlar = {"laptop": 950.0, "fare": 25.0, "monitor": 210.0}

print(siparisi_isle(["laptop", "monitor"], envanter, fiyatlar))
print(siparisi_isle(["fare"], envanter, fiyatlar))
````

## 4. Doğru Yapıyı Seçmek: Arama Hızı Ön İzlemesi
Bir listeyi taramayı sözlükte doğrudan aramayla karşılaştırın — gelecek hafta Big-O ile biçimselleştireceğimiz fikrin aynısı.

````python
import time

# Büyük bir müşteri "tablosunu" iki şekilde oluştur
n = 200_000
musteri_listesi = [{"id": i, "ad": f"Müşteri {i}"} for i in range(n)]
musteri_sozluk = {m["id"]: m for m in musteri_listesi}

hedef_id = n - 1  # doğrusal tarama için en kötü durum id'si

baslangic = time.perf_counter()
for m in musteri_listesi:
    if m["id"] == hedef_id:
        bulunan_dogrusal = m
        break
dogrusal_sure = time.perf_counter() - baslangic

baslangic = time.perf_counter()
bulunan_sozluk = musteri_sozluk[hedef_id]
sozluk_sure = time.perf_counter() - baslangic

print(f"Doğrusal tarama: {dogrusal_sure:.6f}s")
print(f"Sözlük araması: {sozluk_sure:.6f}s")
````
