## 1. Faktöriyel ve Bir Listenin Toplamı

```python
def faktoriyel(n):
    if n == 0:
        return 1
    return n * faktoriyel(n - 1)

def liste_toplami(ogeler):
    if len(ogeler) == 0:
        return 0
    return ogeler[0] + liste_toplami(ogeler[1:])

print(faktoriyel(5))
print(liste_toplami([120, 45, 300, 75]))
```

## 2. Organizasyon Şeması Çalışan Sayısı

```python
org_semasi = {
    "ad": "CEO",
    "raporlar": [
        {"ad": "Satış VP", "raporlar": [
            {"ad": "Satış Temsilcisi A", "raporlar": []},
            {"ad": "Satış Temsilcisi B", "raporlar": []},
        ]},
        {"ad": "Mühendislik VP", "raporlar": [
            {"ad": "Mühendis A", "raporlar": []},
        ]},
    ],
}

def calisan_sayisi(calisan):
    toplam = 1
    for rapor in calisan["raporlar"]:
        toplam += calisan_sayisi(rapor)
    return toplam

print("CEO altında toplam çalışan:", calisan_sayisi(org_semasi))
```

## 3. Fibonacci: Saf ile Memoize Edilmiş

```python
import time

def fib_ozyinelemeli(n):
    if n <= 1:
        return n
    return fib_ozyinelemeli(n - 1) + fib_ozyinelemeli(n - 2)

onbellek = {}
def fib_memo(n):
    if n in onbellek:
        return onbellek[n]
    if n <= 1:
        return n
    sonuc = fib_memo(n - 1) + fib_memo(n - 2)
    onbellek[n] = sonuc
    return sonuc

baslangic = time.perf_counter()
print(fib_ozyinelemeli(28))
print("saf:", time.perf_counter() - baslangic, "s")

baslangic = time.perf_counter()
print(fib_memo(28))
print("memoize edilmiş:", time.perf_counter() - baslangic, "s")
```

## 4. Doğrusal Arama ve İkili Arama

```python
def dogrusal_arama(ogeler, hedef):
    for i, deger in enumerate(ogeler):
        if deger == hedef:
            return i
    return -1

def ikili_arama(ogeler, hedef, alt=0, ust=None):
    if ust is None:
        ust = len(ogeler) - 1
    if alt > ust:
        return -1
    orta = (alt + ust) // 2
    if ogeler[orta] == hedef:
        return orta
    elif ogeler[orta] < hedef:
        return ikili_arama(ogeler, hedef, orta + 1, ust)
    else:
        return ikili_arama(ogeler, hedef, alt, orta - 1)

sirali_idler = list(range(0, 100000, 2))
print(dogrusal_arama(sirali_idler, 87654))
print(ikili_arama(sirali_idler, 87654))
```

## 5. Arama Maliyetlerini Karşılaştırmak

```python
def ikili_arama_sayimli(ogeler, hedef, alt=0, ust=None, sayac=None):
    if sayac is None:
        sayac = [0]
    if ust is None:
        ust = len(ogeler) - 1
    if alt > ust:
        return -1, sayac[0]
    sayac[0] += 1
    orta = (alt + ust) // 2
    if ogeler[orta] == hedef:
        return orta, sayac[0]
    elif ogeler[orta] < hedef:
        return ikili_arama_sayimli(ogeler, hedef, orta + 1, ust, sayac)
    else:
        return ikili_arama_sayimli(ogeler, hedef, alt, orta - 1, sayac)

def dogrusal_arama_sayimli(ogeler, hedef):
    karsilastirma = 0
    for deger in ogeler:
        karsilastirma += 1
        if deger == hedef:
            return karsilastirma
    return karsilastirma

sirali_idler = list(range(0, 500000, 2))
hedef = 499998
_, bsayim = ikili_arama_sayimli(sirali_idler, hedef)
lsayim = dogrusal_arama_sayimli(sirali_idler, hedef)
print("İkili arama karşılaştırmaları:", bsayim)
print("Doğrusal arama karşılaştırmaları:", lsayim)
```

### Çözümler

```python
def geri_sayim(n):
    if n == 0:
        print("Kalkış!")
        return
    print(n)
    geri_sayim(n - 1)

def liste_max(ogeler):
    if len(ogeler) == 1:
        return ogeler[0]
    geri_kalan_max = liste_max(ogeler[1:])
    return ogeler[0] if ogeler[0] > geri_kalan_max else geri_kalan_max

def palindrom_mu(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return palindrom_mu(s[1:-1])

def derin_toplam(veri):
    toplam = 0
    for oge in veri:
        if isinstance(oge, list):
            toplam += derin_toplam(oge)
        else:
            toplam += oge
    return toplam

geri_sayim(3)
print(liste_max([3, 7, 2, 9, 4]))
print(palindrom_mu("ada"), palindrom_mu("merhaba"))
print(derin_toplam([120, [45, 30], [80, [15, 25]], 60]))
```
