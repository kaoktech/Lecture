## 1. Node ve Temel Bir LinkedList

```python
class Node:
    def __init__(self, deger):
        self.deger = deger
        self.sonraki = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, deger):
        yeni_dugum = Node(deger)
        if self.head is None:
            self.head = yeni_dugum
            return
        gecerli = self.head
        while gecerli.sonraki is not None:
            gecerli = gecerli.sonraki
        gecerli.sonraki = yeni_dugum

    def basa_ekle(self, deger):
        yeni_dugum = Node(deger)
        yeni_dugum.sonraki = self.head
        self.head = yeni_dugum

    def sil(self, deger):
        if self.head is None:
            return
        if self.head.deger == deger:
            self.head = self.head.sonraki
            return
        gecerli = self.head
        while gecerli.sonraki is not None:
            if gecerli.sonraki.deger == deger:
                gecerli.sonraki = gecerli.sonraki.sonraki
                return
            gecerli = gecerli.sonraki

    def ara(self, deger):
        gecerli = self.head
        while gecerli is not None:
            if gecerli.deger == deger:
                return True
            gecerli = gecerli.sonraki
        return False

    def listeye_donustur(self):
        degerler = []
        gecerli = self.head
        while gecerli is not None:
            degerler.append(gecerli.deger)
            gecerli = gecerli.sonraki
        return degerler

    def __len__(self):
        sayac = 0
        gecerli = self.head
        while gecerli is not None:
            sayac += 1
            gecerli = gecerli.sonraki
        return sayac
```

```python
siparisler = LinkedList()
siparisler.append("Sipariş #101")
siparisler.append("Sipariş #102")
siparisler.basa_ekle("ACİL Sipariş #999")

print(siparisler.listeye_donustur())
print("Uzunluk:", len(siparisler))
print("Sipariş #102 içeriyor mu:", siparisler.ara("Sipariş #102"))

siparisler.sil("Sipariş #101")
print(siparisler.listeye_donustur())
```

## 2. Zamanlama: Bağlı Listede Başa Ekleme ile Python list'inde

```python
import time

# Bağlı liste başa ekleme: her biri O(1)
bl = LinkedList()
baslangic = time.perf_counter()
for i in range(20000):
    bl.basa_ekle(i)
print("Bağlı liste başa ekleme x20000:", time.perf_counter() - baslangic, "s")

# Python list insert(0, x): her biri O(n)
py_liste = []
baslangic = time.perf_counter()
for i in range(20000):
    py_liste.insert(0, i)
print("list.insert(0, x) x20000:", time.perf_counter() - baslangic, "s")
```

## 3. Çiftli Bağlı Bir Düğüm

```python
class CiftliDugum:
    def __init__(self, deger):
        self.deger = deger
        self.sonraki = None
        self.onceki = None

sayfa1 = CiftliDugum("anasayfa")
sayfa2 = CiftliDugum("urunler")
sayfa1.sonraki = sayfa2
sayfa2.onceki = sayfa1

print(sayfa2.onceki.deger, "->", sayfa2.deger)
```

## 4. İş Senaryosu: Geri Alma Geçmişi

```python
def eylem_ekle(gecmis, eylem):
    gecmis.basa_ekle(eylem)

def geri_al(gecmis):
    if gecmis.head is None:
        return None
    eylem = gecmis.head.deger
    gecmis.head = gecmis.head.sonraki
    return eylem

gecmis = LinkedList()
for eylem in ["A yaz", "B yaz", "B sil", "C yaz"]:
    eylem_ekle(gecmis, eylem)

print("Geçmiş:", gecmis.listeye_donustur())
print("Geri al:", geri_al(gecmis))
print("Geri al:", geri_al(gecmis))
print("Kalan:", gecmis.listeye_donustur())
```

### Çözümler

```python
def sonrasina_ekle(bl, deger, yeni_deger):
    gecerli = bl.head
    while gecerli is not None:
        if gecerli.deger == deger:
            yeni_dugum = Node(yeni_deger)
            yeni_dugum.sonraki = gecerli.sonraki
            gecerli.sonraki = yeni_dugum
            return True
        gecerli = gecerli.sonraki
    return False

def tersine_cevir(bl):
    onceki = None
    gecerli = bl.head
    while gecerli is not None:
        sonraki_dugum = gecerli.sonraki
        gecerli.sonraki = onceki
        onceki = gecerli
        gecerli = sonraki_dugum
    bl.head = onceki

demo = LinkedList()
for v in [1, 2, 3, 4]:
    demo.append(v)
sonrasina_ekle(demo, 2, 99)
print(demo.listeye_donustur())
tersine_cevir(demo)
print(demo.listeye_donustur())
```
