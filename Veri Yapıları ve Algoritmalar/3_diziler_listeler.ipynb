## 2. Karmaşıklık Gösterimi: Sona Ekleme ile Başa Ekleme

```python
import time

n = 200_000

baslangic = time.perf_counter()
veri = []
for i in range(n):
    veri.append(i)          # her seferinde itfa edilmiş O(1)
print("Ekleme (sona)", n, "öğe:", time.perf_counter() - baslangic, "s")

baslangic = time.perf_counter()
veri2 = []
for i in range(2000):        # bilerek çok daha küçük n!
    veri2.insert(0, i)      # her seferinde O(n)
print("Başa ekleme", 2000, "öğe:", time.perf_counter() - baslangic, "s")
```

## 3. Müşteri Kayıtları Listesinde CRUD

```python
musteriler = [{"id": 1, "ad": "Ayşe"}, {"id": 2, "ad": "Mert"}]

# Oluştur
musteriler.append({"id": 3, "ad": "Zeynep"})

# Oku
def musteri_bul(musteriler, mid):
    for m in musteriler:
        if m["id"] == mid:
            return m
    return None

# Güncelle
m = musteri_bul(musteriler, 2)
if m:
    m["ad"] = "Mert Kaya"

# Sil
musteriler = [m for m in musteriler if m["id"] != 1]
print(musteriler)
```

## 4. Liste Comprehension'ları

```python
siparisler = [{"id": 1, "toplam": 250.0}, {"id": 2, "toplam": 980.0}, {"id": 3, "toplam": 45.0}]

buyuk_siparisler = [s for s in siparisler if s["toplam"] > 100]
vergili_toplamlar = [round(s["toplam"] * 1.18, 2) for s in siparisler]

print(buyuk_siparisler)
print(vergili_toplamlar)
```

## 5. 2D Listeler: Bir Satış Tablosu

```python
satis_tablosu = [
    [120, 200, 90],
    [80, 150, 300],
    [60, 40, 55],
]

# Satır toplamları
for i, satir in enumerate(satis_tablosu):
    print(f"Ürün {i}: {sum(satir)}")

# Sütun toplamları
sutun_sayisi = len(satis_tablosu[0])
for sutun in range(sutun_sayisi):
    sutun_toplami = sum(satir[sutun] for satir in satis_tablosu)
    print(f"Ay {sutun}: {sutun_toplami}")
```

```python
envanter = [
    {"ad": "Laptop", "adet": 5, "birim_fiyat": 950.0},
    {"ad": "Fare", "adet": 40, "birim_fiyat": 25.0},
    {"ad": "Monitör", "adet": 8, "birim_fiyat": 210.0},
    {"ad": "Klavye", "adet": 3, "birim_fiyat": 60.0},
    {"ad": "Webcam", "adet": 15, "birim_fiyat": 45.0},
]

az_stok = [i for i in envanter if i["adet"] < 10]
print("Az stok:", [i["ad"] for i in az_stok])

def adet_guncelle(envanter, ad, yeni_adet):
    for item in envanter:
        if item["ad"] == ad:
            item["adet"] = yeni_adet
            return True
    return False

adet_guncelle(envanter, "Fare", 35)

toplam_deger = sum(i["adet"] * i["birim_fiyat"] for i in envanter)
print("Toplam envanter değeri:", toplam_deger)

en_iyi = max(range(len(satis_tablosu)), key=lambda i: max(satis_tablosu[i]))
print("En yüksek tekil aya sahip ürün:", en_iyi)
```
