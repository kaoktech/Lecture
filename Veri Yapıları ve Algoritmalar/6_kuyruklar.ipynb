## 1. list.pop(0) Neden Yavaştır

```python
import time
from collections import deque

n = 20000

py_liste = list(range(n))
baslangic = time.perf_counter()
while py_liste:
    py_liste.pop(0)
print("list.pop(0) x", n, ":", time.perf_counter() - baslangic, "s")

dq = deque(range(n))
baslangic = time.perf_counter()
while dq:
    dq.popleft()
print("deque.popleft() x", n, ":", time.perf_counter() - baslangic, "s")
```

## 2. Kuyruk Sınıfı

```python
from collections import deque

class Kuyruk:
    def __init__(self):
        self._ogeler = deque()

    def enqueue(self, oge):
        self._ogeler.append(oge)

    def dequeue(self):
        return self._ogeler.popleft()

    def front(self):
        return self._ogeler[0]

    def is_empty(self):
        return len(self._ogeler) == 0

    def boyut(self):
        return len(self._ogeler)
```

## 3. Bir Talep Kuyruğunu Simüle Etmek

```python
talepler = Kuyruk()
for t in ["Talep #1", "Talep #2", "Talep #3", "Talep #4"]:
    talepler.enqueue(t)

while not talepler.is_empty():
    print("Şu an işleniyor:", talepler.dequeue())
```

## 5. Bir Deque ile Kayan Pencere

```python
def okuma_ekle(pencere, deger, max_boyut):
    pencere.append(deger)
    if len(pencere) > max_boyut:
        pencere.popleft()

pencere = deque()
for v in range(1, 11):
    okuma_ekle(pencere, v, max_boyut=5)
    print(list(pencere))

print("Mevcut pencerenin ortalaması:", sum(pencere) / len(pencere))
```

```python
def birikmis_mi(kuyruk):
    return kuyruk.boyut() > 3

k = Kuyruk()
for i in range(5):
    k.enqueue(i)
print("Birikmiş:", birikmis_mi(k))

def pencereli_ortalama(degerler, max_boyut):
    pencere = deque()
    ortalamalar = []
    for v in degerler:
        pencere.append(v)
        if len(pencere) > max_boyut:
            pencere.popleft()
        ortalamalar.append(sum(pencere) / len(pencere))
    return ortalamalar

print(pencereli_ortalama([10, 20, 30, 40, 50, 60], max_boyut=3))
```
