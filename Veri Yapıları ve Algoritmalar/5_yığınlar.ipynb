## 1. Yigin Sınıfı

```python
class Yigin:
    def __init__(self):
        self._ogeler = []

    def push(self, oge):
        self._ogeler.append(oge)

    def pop(self):
        return self._ogeler.pop()

    def peek(self):
        return self._ogeler[-1]

    def is_empty(self):
        return len(self._ogeler) == 0

    def __len__(self):
        return len(self._ogeler)
```

## 2. Geri Alma Özelliği

```python
geri_al_yigini = Yigin()

def eylem_gerceklestir(eylem):
    print("Yapılıyor:", eylem)
    geri_al_yigini.push(eylem)

def geri_al():
    if geri_al_yigini.is_empty():
        print("Geri alınacak bir şey yok")
        return
    eylem = geri_al_yigini.pop()
    print("Geri alınıyor:", eylem)

eylem_gerceklestir("'Merhaba' yaz")
eylem_gerceklestir("seçimi kalınlaştır")
geri_al()
geri_al()
geri_al()  # geriye bir şey kalmadı
```

## 3. Dengeli Parantezler

```python
def dengeli_mi(metin):
    yigin = Yigin()
    ciftler = {")": "(", "]": "[", "}": "{"}
    for karakter in metin:
        if karakter in "([{":
            yigin.push(karakter)
        elif karakter in ")]}":
            if yigin.is_empty() or yigin.pop() != ciftler[karakter]:
                return False
    return yigin.is_empty()

testler = ["( [ ] )", "( [ ) ]", "{[()()]}", "((("]
for t in testler:
    print(repr(t), "->", dengeli_mi(t))
```

## 4. Bir String'i Tersine Çevirmek

```python
def string_tersine_cevir(s):
    yigin = Yigin()
    for karakter in s:
        yigin.push(karakter)
    tersine_cevrilmis = []
    while not yigin.is_empty():
        tersine_cevrilmis.append(yigin.pop())
    return "".join(tersine_cevrilmis)

print(string_tersine_cevir("fatura"))
```

## 5. Sonek İfade Değerlendirmesi

```python
def sonek_degerlendir(ifade):
    yigin = Yigin()
    for token in ifade.split():
        if token in "+-*/":
            b = yigin.pop()
            a = yigin.pop()
            if token == "+": yigin.push(a + b)
            elif token == "-": yigin.push(a - b)
            elif token == "*": yigin.push(a * b)
            elif token == "/": yigin.push(a / b)
        else:
            yigin.push(float(token))
    return yigin.pop()

print(sonek_degerlendir("3 4 2 * +"))
print(sonek_degerlendir("10 2 /"))
```
