## 1. O(1), O(n), O(n²): Timing Them Directly

```python
import time

def get_first(items):
    return items[0]

def linear_search(items, target):
    for x in items:
        if x == target:
            return True
    return False

def has_duplicate_pair(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return True
    return False
```

```python
sizes = [1000, 2000, 4000]
for n in sizes:
    data = list(range(n))

    start = time.perf_counter()
    get_first(data)
    t_const = time.perf_counter() - start

    start = time.perf_counter()
    linear_search(data, -1)  # worst case: not found
    t_linear = time.perf_counter() - start

    start = time.perf_counter()
    has_duplicate_pair(data)  # worst case: no duplicates
    t_quad = time.perf_counter() - start

    print(f"n={n:>5} | O(1): {t_const:.6f}s | O(n): {t_linear:.6f}s | O(n^2): {t_quad:.6f}s")
```

Notice how the O(1) time barely changes, O(n) time grows roughly proportionally to n, and O(n²) time grows much faster — doubling n roughly quadruples it.

## 2. Business Scenario: Two Sales-Report Algorithms
Compare Developer A's O(n²)-ish approach with Developer B's O(n) approach.

```python
import random

random.seed(0)
products = [f"P{i}" for i in range(50)]
transactions = [{"product": random.choice(products), "amount": round(random.uniform(10, 500), 2)}
                for _ in range(20000)]

def developer_a(transactions, products):
    totals = {}
    for prod in products:                     # p products
        total = 0
        for t in transactions:                # n transactions, scanned per product
            if t["product"] == prod:
                total += t["amount"]
        totals[prod] = total
    return totals

def developer_b(transactions):
    totals = {}
    for t in transactions:                     # one pass, O(n)
        totals[t["product"]] = totals.get(t["product"], 0) + t["amount"]
    return totals

start = time.perf_counter(); developer_a(transactions, products); print("Developer A:", time.perf_counter() - start, "s")
start = time.perf_counter(); developer_b(transactions); print("Developer B:", time.perf_counter() - start, "s")
```

## 3. Binary Search — O(log n)

```python
def binary_search(sorted_ids, target):
    low, high = 0, len(sorted_ids) - 1
    steps = 0
    while low <= high:
        steps += 1
        mid = (low + high) // 2
        if sorted_ids[mid] == target:
            return mid, steps
        elif sorted_ids[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, steps

sorted_ids = list(range(0, 1_000_000, 2))  # 500,000 items
index, steps = binary_search(sorted_ids, 999998)
print(f"Found at index {index} in only {steps} steps out of 500,000 items")
```

## 4. Time–Space Trade-off: Recompute vs. Index

```python
def total_sales_recompute(transactions, product):
    return sum(t["amount"] for t in transactions if t["product"] == product)

def build_sales_index(transactions):
    index = {}
    for t in transactions:
        index[t["product"]] = index.get(t["product"], 0) + t["amount"]
    return index

# Recompute: cheap in memory, expensive if called many times
start = time.perf_counter()
for _ in range(20):
    total_sales_recompute(transactions, "P1")
print("20 recomputes:", time.perf_counter() - start, "s")

# Index once, then O(1) lookups
start = time.perf_counter()
index = build_sales_index(transactions)
for _ in range(20):
    index["P1"]
print("Build once + 20 lookups:", time.perf_counter() - start, "s")
```
