## 2. Complexity Demo: Append vs. Insert at Front

```python
import time

n = 200_000

start = time.perf_counter()
data = []
for i in range(n):
    data.append(i)          # O(1) amortized each time
print("Appending", n, "items:", time.perf_counter() - start, "s")

start = time.perf_counter()
data2 = []
for i in range(2000):        # much smaller n on purpose!
    data2.insert(0, i)      # O(n) each time
print("Inserting at front", 2000, "items:", time.perf_counter() - start, "s")
```

## 3. CRUD on a List of Customer Records

```python
customers = [{"id": 1, "name": "Ayşe"}, {"id": 2, "name": "Mert"}]

# Create
customers.append({"id": 3, "name": "Zeynep"})

# Read
def find_customer(customers, cid):
    for c in customers:
        if c["id"] == cid:
            return c
    return None

# Update
c = find_customer(customers, 2)
if c:
    c["name"] = "Mert Kaya"

# Delete
customers = [c for c in customers if c["id"] != 1]
print(customers)
```

## 4. List Comprehensions

```python
orders = [{"id": 1, "total": 250.0}, {"id": 2, "total": 980.0}, {"id": 3, "total": 45.0}]

big_orders = [o for o in orders if o["total"] > 100]
totals_with_tax = [round(o["total"] * 1.18, 2) for o in orders]

print(big_orders)
print(totals_with_tax)
```

## 5. 2D Lists: A Sales Table

```python
sales_table = [
    [120, 200, 90],
    [80, 150, 300],
    [60, 40, 55],
]

# Row totals
for i, row in enumerate(sales_table):
    print(f"Product {i}: {sum(row)}")

# Column totals
num_cols = len(sales_table[0])
for col in range(num_cols):
    col_total = sum(row[col] for row in sales_table)
    print(f"Month {col}: {col_total}")
```

## 6. Copies vs. References

```python
original = [1, 2, 3]
alias = original
copy = original[:]

alias.append(4)
copy.append(99)

print("original:", original)
print("alias:", alias)
print("copy:", copy)
```
