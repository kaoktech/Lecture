## 2. Business Scenario: Loan Pre-Approval Algorithm
Follow the five properties of a good algorithm: clear input, clear output, precise steps, it terminates, and every step is executable.

````python
def preapprove_loan(income, existing_debt, credit_score):
    debt_to_income = existing_debt / income
    if credit_score >= 700 and debt_to_income < 0.35:
        return "approved"
    elif credit_score >= 600 and debt_to_income < 0.45:
        return "needs manual review"
    else:
        return "rejected"

print(preapprove_loan(60000, 15000, 720))
print(preapprove_loan(50000, 20000, 650))
print(preapprove_loan(30000, 20000, 550))
````

## 3. Decomposition: Processing an Order
Break "process this order" into validate → calculate → decide, each as its own function.

````python
def validate_order(cart, inventory):
    return all(inventory.get(item, 0) > 0 for item in cart)

def calculate_total(cart, prices, tax_rate=0.18):
    subtotal = sum(prices[item] for item in cart)
    return subtotal * (1 + tax_rate)

def process_order(cart, inventory, prices):
    if not validate_order(cart, inventory):
        return "Order rejected: item out of stock"
    return round(calculate_total(cart, prices), 2)

inventory = {"laptop": 5, "mouse": 0, "monitor": 12}
prices = {"laptop": 950.0, "mouse": 25.0, "monitor": 210.0}

print(process_order(["laptop", "monitor"], inventory, prices))
print(process_order(["mouse"], inventory, prices))
````

## 4. Choosing the Right Structure: Search Speed Preview
Compare scanning a list versus looking up directly in a dictionary — the same idea we'll formalize with Big-O next week.

````python
import time

# Build a large customer "table" two ways
n = 200_000
customer_list = [{"id": i, "name": f"Customer {i}"} for i in range(n)]
customer_dict = {c["id"]: c for c in customer_list}

target_id = n - 1  # a worst-case id for the linear scan

start = time.perf_counter()
for c in customer_list:
    if c["id"] == target_id:
        found_linear = c
        break
linear_time = time.perf_counter() - start

start = time.perf_counter()
found_dict = customer_dict[target_id]
dict_time = time.perf_counter() - start

print(f"Linear scan: {linear_time:.6f}s")
print(f"Dictionary lookup: {dict_time:.6f}s")
````
