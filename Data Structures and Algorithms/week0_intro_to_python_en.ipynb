## 2. Variables and Types
A variable is a name bound to a value. Python figures out the type automatically.

```python
customer_name = "Ayşe Demir"
order_total = 249.90
is_vip = True

print(customer_name, order_total, is_vip)
print(type(customer_name), type(order_total), type(is_vip))
```

```python
# Converting between types (casting)
price_text = "19.99"
print(type(price_text))

price = float(price_text)
tax = price * 0.18
print(type(price), round(price + tax, 2))
```

## 3. Operators
Arithmetic: `+ - * / // % **`  |  Comparison: `== != > < >= <=`  |  Logical: `and or not`

```python
units = 12
unit_price = 45.0
discount_rate = 0.10 if units >= 10 else 0.0

subtotal = units * unit_price
total = subtotal * (1 - discount_rate)
print("Subtotal:", subtotal, "| Total after discount:", total)
```

## 4. Conditionals — `if` / `elif` / `else`
**Business scenario:** decide a loyalty discount from membership tier and order size.

```python
def discount_rate(tier, order_total):
    if tier == "gold":
        return 0.15
    elif tier == "silver" and order_total > 500:
        return 0.10
    else:
        return 0.0

print(discount_rate("silver", 620))
print(discount_rate("bronze", 620))
```

## 5. Loops — `for` and `while`
**Business scenario:** total a batch of orders and flag any order above a risk threshold.

```python
# for: iterate a known collection
orders = [120.50, 980.00, 45.25, 1200.00, 300.75]
running_total = 0

for amount in orders:
    running_total += amount
    flag = " <-- HIGH VALUE, review" if amount > 1000 else ""
    print(f"Order: {amount:>8.2f} | Running total: {running_total:>9.2f}{flag}")
```

```python
# while: repeat until a condition becomes False
attempts = 0
while attempts < 3:
    print("Login attempt", attempts + 1)
    attempts += 1
```

## 6. Core Data Containers — list, tuple, dict, set

```python
# list — ordered, mutable
inventory = [120, 45, 0, 300]
inventory.append(75)
inventory[2] = 10
print(inventory, inventory[1:3])
```

```python
# tuple — ordered, immutable
coordinates = (41.015, 28.979)  # (lat, lon)
print(coordinates[0])

# set — unordered, unique
skus_sold_today = {"A1", "B2", "A1", "C3"}
print(skus_sold_today)
```

```python
# dict — key/value lookup
customer = {"id": 501, "name": "Mert Kaya", "vip": True}
print(customer["name"])
customer["email"] = "mert@example.com"
print(customer)
```

## 7. Functions
Package logic into a reusable, named block. Parameters are inputs; `return` sends back a result.

```python
def invoice_total(unit_price, quantity, tax_rate=0.18):
    subtotal = unit_price * quantity
    tax = subtotal * tax_rate
    return subtotal + tax

print(invoice_total(150, 3))
print(invoice_total(150, 3, tax_rate=0.08))
```
