## 1. The Stack Class

```python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        return self._items.pop()

    def peek(self):
        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def __len__(self):
        return len(self._items)
```

## 2. Undo Feature

```python
undo_stack = Stack()

def perform_action(action):
    print("Doing:", action)
    undo_stack.push(action)

def undo():
    if undo_stack.is_empty():
        print("Nothing to undo")
        return
    action = undo_stack.pop()
    print("Undoing:", action)

perform_action("type 'Hello'")
perform_action("bold selection")
undo()
undo()
undo()  # nothing left
```

## 3. Balanced Brackets

```python
def is_balanced(text):
    stack = Stack()
    pairs = {")": "(", "]": "[", "}": "{"}
    for char in text:
        if char in "([{":
            stack.push(char)
        elif char in ")]}":
            if stack.is_empty() or stack.pop() != pairs[char]:
                return False
    return stack.is_empty()

tests = ["( [ ] )", "( [ ) ]", "{[()()]}", "((("]
for t in tests:
    print(repr(t), "->", is_balanced(t))
```

## 4. Reversing a String

```python
def reverse_string(s):
    stack = Stack()
    for char in s:
        stack.push(char)
    reversed_chars = []
    while not stack.is_empty():
        reversed_chars.append(stack.pop())
    return "".join(reversed_chars)

print(reverse_string("invoice"))
```

## 5. Postfix Expression Evaluation

```python
def evaluate_postfix(expression):
    stack = Stack()
    for token in expression.split():
        if token in "+-*/":
            b = stack.pop()
            a = stack.pop()
            if token == "+": stack.push(a + b)
            elif token == "-": stack.push(a - b)
            elif token == "*": stack.push(a * b)
            elif token == "/": stack.push(a / b)
        else:
            stack.push(float(token))
    return stack.pop()

print(evaluate_postfix("3 4 2 * +"))
print(evaluate_postfix("10 2 /"))
```

```python
def evaluate_postfix_v2(expression):
    stack = Stack()
    for token in expression.split():
        if token in "+-*/%":
            b = stack.pop()
            a = stack.pop()
            if token == "+": stack.push(a + b)
            elif token == "-": stack.push(a - b)
            elif token == "*": stack.push(a * b)
            elif token == "%": stack.push(a % b)
            elif token == "/":
                if b == 0:
                    return "Error: division by zero"
                stack.push(a / b)
        else:
            stack.push(float(token))
    return stack.pop()

print(evaluate_postfix_v2("10 2 /"))
print(evaluate_postfix_v2("5 0 /"))
print(evaluate_postfix_v2("6 3 %"))
```
