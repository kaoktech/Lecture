## 1. Why list.pop(0) Is Slow

```python
import time
from collections import deque

n = 20000

py_list = list(range(n))
start = time.perf_counter()
while py_list:
    py_list.pop(0)
print("list.pop(0) x", n, ":", time.perf_counter() - start, "s")

dq = deque(range(n))
start = time.perf_counter()
while dq:
    dq.popleft()
print("deque.popleft() x", n, ":", time.perf_counter() - start, "s")
```

## 2. The Queue Class

```python
from collections import deque

class Queue:
    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        self._items.append(item)

    def dequeue(self):
        return self._items.popleft()

    def front(self):
        return self._items[0]

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)
```

## 3. Simulating a Ticket Queue

```python
tickets = Queue()
for t in ["Ticket #1", "Ticket #2", "Ticket #3", "Ticket #4"]:
    tickets.enqueue(t)

while not tickets.is_empty():
    print("Now serving:", tickets.dequeue())
```

## 5. Sliding Window with a Deque

```python
def add_reading(window, value, max_size):
    window.append(value)
    if len(window) > max_size:
        window.popleft()

window = deque()
for v in range(1, 11):
    add_reading(window, v, max_size=5)
    print(list(window))

print("Average of current window:", sum(window) / len(window))
```

### Solutions

```python
def is_backed_up(queue):
    return queue.size() > 3

q = Queue()
for i in range(5):
    q.enqueue(i)
print("Backed up:", is_backed_up(q))

def windowed_average(values, max_size):
    window = deque()
    averages = []
    for v in values:
        window.append(v)
        if len(window) > max_size:
            window.popleft()
        averages.append(sum(window) / len(window))
    return averages

print(windowed_average([10, 20, 30, 40, 50, 60], max_size=3))
```
