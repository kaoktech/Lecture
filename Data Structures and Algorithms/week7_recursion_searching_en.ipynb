## 1. Factorial and Sum of a List

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

def sum_list(items):
    if len(items) == 0:
        return 0
    return items[0] + sum_list(items[1:])

print(factorial(5))
print(sum_list([120, 45, 300, 75]))
```

## 2. Org Chart Headcount

```python
org_chart = {
    "name": "CEO",
    "reports": [
        {"name": "VP Sales", "reports": [
            {"name": "Sales Rep A", "reports": []},
            {"name": "Sales Rep B", "reports": []},
        ]},
        {"name": "VP Eng", "reports": [
            {"name": "Engineer A", "reports": []},
        ]},
    ],
}

def headcount(employee):
    total = 1
    for report in employee["reports"]:
        total += headcount(report)
    return total

print("Total headcount under CEO:", headcount(org_chart))
```

## 3. Fibonacci: Naive vs. Memoized

```python
import time

def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)

cache = {}
def fib_memo(n):
    if n in cache:
        return cache[n]
    if n <= 1:
        return n
    result = fib_memo(n - 1) + fib_memo(n - 2)
    cache[n] = result
    return result

start = time.perf_counter()
print(fib_recursive(28))
print("naive:", time.perf_counter() - start, "s")

start = time.perf_counter()
print(fib_memo(28))
print("memoized:", time.perf_counter() - start, "s")
```

## 4. Linear Search and Binary Search

```python
def linear_search(items, target):
    for i, value in enumerate(items):
        if value == target:
            return i
    return -1

def binary_search(items, target, low=0, high=None):
    if high is None:
        high = len(items) - 1
    if low > high:
        return -1
    mid = (low + high) // 2
    if items[mid] == target:
        return mid
    elif items[mid] < target:
        return binary_search(items, target, mid + 1, high)
    else:
        return binary_search(items, target, low, mid - 1)

sorted_ids = list(range(0, 100000, 2))
print(linear_search(sorted_ids, 87654))
print(binary_search(sorted_ids, 87654))
```

## 5. Comparing Search Costs

```python
def binary_search_counted(items, target, low=0, high=None, count=None):
    if count is None:
        count = [0]
    if high is None:
        high = len(items) - 1
    if low > high:
        return -1, count[0]
    count[0] += 1
    mid = (low + high) // 2
    if items[mid] == target:
        return mid, count[0]
    elif items[mid] < target:
        return binary_search_counted(items, target, mid + 1, high, count)
    else:
        return binary_search_counted(items, target, low, mid - 1, count)

def linear_search_counted(items, target):
    comparisons = 0
    for value in items:
        comparisons += 1
        if value == target:
            return comparisons
    return comparisons

sorted_ids = list(range(0, 500000, 2))
target = 499998
_, bcount = binary_search_counted(sorted_ids, target)
lcount = linear_search_counted(sorted_ids, target)
print("Binary search comparisons:", bcount)
print("Linear search comparisons:", lcount)
```

### Solutions

```python
def count_down(n):
    if n == 0:
        print("Liftoff!")
        return
    print(n)
    count_down(n - 1)

def max_in_list(items):
    if len(items) == 1:
        return items[0]
    rest_max = max_in_list(items[1:])
    return items[0] if items[0] > rest_max else rest_max

def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])

def deep_sum(data):
    total = 0
    for item in data:
        if isinstance(item, list):
            total += deep_sum(item)
        else:
            total += item
    return total

count_down(3)
print(max_in_list([3, 7, 2, 9, 4]))
print(is_palindrome("madam"), is_palindrome("hello"))
print(deep_sum([120, [45, 30], [80, [15, 25]], 60]))
```
