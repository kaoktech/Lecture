## Part 1 — Basic try/except

```python
def get_average(numbers):
    try:
        return sum(numbers) / len(numbers)
    except ZeroDivisionError:
        print("Can't average an empty list.")
        return 0.0


print(get_average([90, 80, 70]))
print(get_average([]))
```

## Part 2 — Custom Exceptions

```python
class InsufficientFundsError(Exception):
    pass

class InvalidPinError(Exception):
    pass


class ATM:
    def __init__(self, correct_pin, balance):
        self._correct_pin = correct_pin
        self.balance = balance

    def withdraw(self, pin, amount):
        if pin != self._correct_pin:
            raise InvalidPinError("Incorrect PIN.")
        if amount > self.balance:
            raise InsufficientFundsError("Insufficient balance.")
        self.balance -= amount
        return amount


atm = ATM(correct_pin="1234", balance=500)

for pin, amount in [("0000", 100), ("1234", 1000), ("1234", 200)]:
    try:
        atm.withdraw(pin, amount)
        print(f"Dispensed {amount}. Remaining: {atm.balance}")
    except InvalidPinError as e:
        print("Denied:", e)
    except InsufficientFundsError as e:
        print("Denied:", e)
```

## Part 3 — finally for Cleanup

```python
def process_transaction(atm, pin, amount):
    print("--- starting transaction ---")
    try:
        atm.withdraw(pin, amount)
        print("Success.")
    except (InvalidPinError, InsufficientFundsError) as e:
        print("Failed:", e)
    finally:
        print("--- transaction attempt complete ---")


process_transaction(atm, "1234", 50)
process_transaction(atm, "0000", 50)
```
