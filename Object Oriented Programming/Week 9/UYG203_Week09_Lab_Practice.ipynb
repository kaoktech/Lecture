## Part 1 — Composition: Student and Its Transcript

```python
class Student:
    def __init__(self, name):
        self.name = name
        self._transcript = []   # composition -- owned by this Student

    def add_grade(self, course, grade):
        self._transcript.append((course, grade))

    def gpa(self):
        if not self._transcript:
            return 0.0
        return sum(g for _, g in self._transcript) / len(self._transcript)


s = Student("Ada")
s.add_grade("UYG203", 92)
s.add_grade("MAT101", 85)
print(s.gpa())
```

## Part 2 — Aggregation: Instructor and Department

```python
class Instructor:
    def __init__(self, name, specialty):
        self.name = name
        self.specialty = specialty


class Department:
    def __init__(self, name):
        self.name = name
        self.instructors = []   # aggregation
        self.students = []      # aggregation

    def add_instructor(self, instructor):
        self.instructors.append(instructor)

    def add_student(self, student):
        self.students.append(student)

    def average_gpa(self):
        if not self.students:
            return 0.0
        return sum(st.gpa() for st in self.students) / len(self.students)


cs = Department("Computer Science")
cs.add_instructor(Instructor("Dr. Ada", "Algorithms"))
cs.add_student(s)
print(cs.average_gpa())
```

## Part 3 — A Third Class: Course

```python
class Course:
    def __init__(self, code, instructor):
        self.code = code
        self.instructor = instructor    # aggregation
        self.enrolled_students = []     # aggregation

    def enroll(self, student):
        self.enrolled_students.append(student)
        print(f"{student.name} enrolled in {self.code}")


uyg203 = Course("UYG203", cs.instructors[0])
uyg203.enroll(s)
```
