## Part 1 — Rebuilding `Student` as a Class

In Week 1 we used a tuple `(name, grade, attendance)` plus standalone functions.
Let's rebuild that exact example as a class.


```python
class Student:
    def __init__(self, name, grade, attendance):
        self.name = name
        self.grade = grade
        self.attendance = attendance

    def is_passing(self):
        if self.attendance < 0.70:
            return False
        return self.grade >= 60

    def summary(self):
        return f"{self.name}: grade={self.grade}, attendance={self.attendance:.0%}"


ada = Student("Ada", 92, 0.95)
mert = Student("Mert", 84, 0.65)

print(ada.summary(), "-> passing:", ada.is_passing())
print(mert.summary(), "-> passing:", mert.is_passing())
```

## Part 2 — Instance Attributes vs. Class Attributes

Let's confirm the distinction from the slides with a quick experiment.


```python
class StudentV2:
    school = "Beykent University"   # class attribute -- shared
    total_students = 0              # class attribute -- shared counter

    def __init__(self, name):
        self.name = name            # instance attribute -- unique per object
        StudentV2.total_students += 1


a = StudentV2("Ada")
b = StudentV2("Mert")

print(a.school, b.school)                 # same value, shared
print(a.name, b.name)                     # different values
print("Total students created:", StudentV2.total_students)
```

### The mutable default argument trap

Run both cells below and compare the output. This is one of the most common
real-world Python bugs.


```python
class StudentBad:
    def __init__(self, name, grades=[]):   # DON'T: mutable default
        self.name = name
        self.grades = grades

x = StudentBad("Ada")
x.grades.append(92)
y = StudentBad("Mert")
print(y.grades)  # unexpectedly [92] !
```

```python
class StudentGood:
    def __init__(self, name, grades=None):  # DO: default to None
        self.name = name
        self.grades = grades if grades else []

x = StudentGood("Ada")
x.grades.append(92)
y = StudentGood("Mert")
print(y.grades)  # correctly []
```

## Part 4 — The `Book` Class

Attributes: `title`, `author`, `isbn`, `is_checked_out`
Methods: `checkout()`, `return_book()`


```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_checked_out = False

    def checkout(self):
        if self.is_checked_out:
            print(f"{self.title} is already checked out.")
        else:
            self.is_checked_out = True
            print(f"You checked out {self.title}.")

    def return_book(self):
        self.is_checked_out = False
        print(f"{self.title} has been returned.")


b1 = Book("Clean Code", "Robert C. Martin", "001")
b1.checkout()
b1.checkout()   # should print "already checked out"
b1.return_book()
```

## Part 5 — The `Product` Class

Attributes: `name`, `price`, `quantity_in_stock`
Methods: `apply_discount(percent)`, `total_value()`


```python
class Product:
    def __init__(self, name, price, quantity_in_stock):
        self.name = name
        self.price = price
        self.quantity_in_stock = quantity_in_stock

    def apply_discount(self, percent):
        self.price = self.price * (1 - percent / 100)

    def total_value(self):
        return self.price * self.quantity_in_stock


p1 = Product("Wireless Mouse", 250.0, 40)
print("Before discount:", p1.total_value())
p1.apply_discount(10)
print("After 10% discount:", p1.total_value())
```
