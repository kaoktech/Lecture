```python
# B3: predict the output before running
class Account:
    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Negative balance!")
        self._balance = value

acc = Account(100)
try:
    acc.balance = -50
except ValueError as e:
    print("Caught:", e)
print(acc.balance)
```

### Specification: A Library Membership System

Build the following, using everything from Weeks 1–7:

1. An abstract base class `LibraryMember(ABC)` with:
   - `__init__(self, name)` storing `name`
   - A protected `_books_borrowed` list, starting empty
   - An abstract method `borrowing_limit(self)` returning the max books this member type can borrow
   - A concrete method `borrow_book(self, title)` that adds `title` to `_books_borrowed`
     **only if** the member hasn't reached their `borrowing_limit()` — otherwise print a message
   - A concrete `__str__` returning something like `"Ada (Student): 2/3 books"`

2. Two subclasses:
   - `Student(LibraryMember)` — `borrowing_limit()` returns `3`
   - `Faculty(LibraryMember)` — `borrowing_limit()` returns `10`

3. A validated `email` property on `LibraryMember` (must contain `"@"`)

Build it step by step below.


```python
from abc import ABC, abstractmethod

class LibraryMember(ABC):
    def __init__(self, name, email):
        self.name = name
        self.email = email          # goes through the property setter
        self._books_borrowed = []

    @property
    def email(self):
        return self._email

    @email.setter
    def email(self, value):
        if "@" not in value:
            raise ValueError("Invalid email address.")
        self._email = value

    @abstractmethod
    def borrowing_limit(self):
        pass

    def borrow_book(self, title):
        if len(self._books_borrowed) >= self.borrowing_limit():
            print(f"{self.name} has reached their borrowing limit.")
            return
        self._books_borrowed.append(title)

    def __str__(self):
        return f"{self.name} ({self.__class__.__name__}): {len(self._books_borrowed)}/{self.borrowing_limit()} books"
```
