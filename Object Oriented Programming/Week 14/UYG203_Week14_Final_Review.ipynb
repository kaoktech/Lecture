## Part 1 — The Capstone Example

This combines abstraction, encapsulation, inheritance, polymorphism, a custom
exception, and aggregation in one small system. Read it, run it, and make sure
you can explain every line before presenting your own project.


```python
from abc import ABC, abstractmethod

class InsufficientFundsError(Exception):
    """Raised when a withdrawal would exceed the account balance."""
    pass


class Account(ABC):
    """Abstract base for any kind of bank account."""

    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance          # Encapsulation: protected attribute

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value

    @abstractmethod
    def account_type(self):              # Abstraction: required in every subclass
        pass

    def __str__(self):
        return f"{self.owner}'s {self.account_type()} account: {self.balance}"
```

```python
class SavingsAccount(Account):        # Inheritance
    def account_type(self):               # Polymorphism: overridden per subclass
        return "Savings"

    def withdraw(self, amount):
        if amount > self._balance:
            raise InsufficientFundsError(f"Cannot withdraw {amount}, balance is {self._balance}")
        self._balance -= amount


class CheckingAccount(Account):           # Inheritance
    def account_type(self):               # Polymorphism
        return "Checking"

    def withdraw(self, amount):
        overdraft_limit = 100
        if amount > self._balance + overdraft_limit:
            raise InsufficientFundsError("Exceeds overdraft limit.")
        self._balance -= amount
```

```python
class Bank:                            # Aggregation: holds accounts it didn't create
    def __init__(self, name):
        self.name = name
        self.accounts = []

    def add_account(self, account):
        self.accounts.append(account)

    def total_assets(self):
        return sum(a.balance for a in self.accounts)


bank = Bank("UYG Bank")
bank.add_account(SavingsAccount("Ada", 1000))
bank.add_account(CheckingAccount("Mert", 200))

for acc in bank.accounts:
    print(acc)

print("Total assets:", bank.total_assets())
```

```python
# Polymorphism in the wild: same method call, different behavior per account type
for acc in bank.accounts:
    try:
        acc.withdraw(5000)
    except InsufficientFundsError as e:
        print(f"{acc.owner}: {e}")
```
