## Part 1 — From Getters/Setters to @property

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value


acc = BankAccount("Ada", 1000)
print(acc.balance)
acc.balance = 2000
print(acc.balance)
````

## Part 2 — A Read-Only Computed Property

````python
class Student:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"


s = Student("Ada", "Yilmaz")
print(s.full_name)
````
