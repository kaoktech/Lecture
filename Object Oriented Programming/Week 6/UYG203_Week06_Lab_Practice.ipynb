## Part 1 — The Payment Classes

```python
class PaymentMethod:
    def pay(self, amount):
        raise NotImplementedError("Subclasses must implement pay()")

    def __str__(self):
        return self.__class__.__name__


class CreditCard(PaymentMethod):
    def __init__(self, last_four):
        self.last_four = last_four

    def pay(self, amount):
        print(f"Charged {amount} to card ending in {self.last_four}.")

    def __str__(self):
        return f"CreditCard(**** {self.last_four})"


class PayPal(PaymentMethod):
    def __init__(self, email):
        self.email = email

    def pay(self, amount):
        print(f"Paid {amount} via PayPal ({self.email}).")

    def __str__(self):
        return f"PayPal({self.email})"


class Cash(PaymentMethod):
    def pay(self, amount):
        print(f"Received {amount} in cash.")

    def __str__(self):
        return "Cash"
```

## Part 2 — Checkout Code That Doesn't Care

```python
def checkout(payment_method, amount):
    print(f"Processing via {payment_method}...")
    payment_method.pay(amount)


orders = [
    (CreditCard("4242"), 250),
    (PayPal("ada@example.com"), 80),
    (Cash(), 40),
]

for method, amount in orders:
    checkout(method, amount)
```

## Part 3 — Duck Typing Without a Shared Parent

```python
class GiftCard:                 # NOT a PaymentMethod subclass at all!
    def __init__(self, code, balance):
        self.code = code
        self.balance = balance

    def pay(self, amount):
        if amount > self.balance:
            print("Insufficient gift card balance.")
            return
        self.balance -= amount
        print(f"Paid {amount} with gift card. Remaining: {self.balance}")

    def __str__(self):
        return f"GiftCard({self.code})"


checkout(GiftCard("GC-001", 100), 30)   # works, thanks to duck typing!
```
