## Part 1 — The Vehicle Parent Class

```python
class Vehicle:
    def __init__(self, brand, model, fuel_level=100):
        self.brand = brand
        self.model = model
        self.fuel_level = fuel_level

    def describe(self):
        return f"{self.brand} {self.model}"

    def refuel(self, amount):
        self.fuel_level = min(100, self.fuel_level + amount)


v = Vehicle("Generic", "Vehicle")
print(v.describe())
```

## Part 2 — Car(Vehicle)

```python
class Car(Vehicle):
    def __init__(self, brand, model, num_doors, fuel_level=100):
        super().__init__(brand, model, fuel_level)
        self.num_doors = num_doors

    def describe(self):
        base = super().describe()
        return f"{base} ({self.num_doors}-door)"


my_car = Car("Toyota", "Corolla", 4)
print(my_car.describe())
print(isinstance(my_car, Vehicle))
```

## Part 3 — Motorcycle(Vehicle)

```python
class Motorcycle(Vehicle):
    def __init__(self, brand, model, has_sidecar=False, fuel_level=100):
        super().__init__(brand, model, fuel_level)
        self.has_sidecar = has_sidecar

    def describe(self):
        extra = " with sidecar" if self.has_sidecar else ""
        return f"{super().describe()}{extra}"


bike = Motorcycle("Honda", "CB500", has_sidecar=True)
print(bike.describe())
```
