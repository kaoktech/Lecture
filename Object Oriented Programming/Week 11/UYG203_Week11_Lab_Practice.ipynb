## Part 2 — CSV

```python
import csv

students = [
    {"name": "Ada", "grade": 92},
    {"name": "Mert", "grade": 84},
]

with open("grades.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "grade"])
    writer.writeheader()
    writer.writerows(students)

with open("grades.csv", "r", newline="") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row["name"], row["grade"], type(row["grade"]))
```

## Part 3 — JSON and Object Serialization

```python
import json

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def to_dict(self):
        return {"name": self.name, "grades": self.grades}

    @classmethod
    def from_dict(cls, data):
        return cls(data["name"], data["grades"])

    def average(self):
        return sum(self.grades) / len(self.grades) if self.grades else 0.0


ada = Student("Ada", [92, 85, 78])

with open("ada.json", "w") as file:
    json.dump(ada.to_dict(), file, indent=2)

with open("ada.json", "r") as file:
    restored = Student.from_dict(json.load(file))

print(restored.name, restored.grades, restored.average())
```

## Part 4 — The GradeManager Class

```python
class GradeManager:
    def __init__(self, filename="grades_manager.json"):
        self.filename = filename
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def save(self):
        with open(self.filename, "w") as file:
            json.dump([s.to_dict() for s in self.students], file, indent=2)

    def load(self):
        try:
            with open(self.filename, "r") as file:
                data = json.load(file)
            self.students = [Student.from_dict(d) for d in data]
        except FileNotFoundError:
            self.students = []


manager = GradeManager()
manager.add_student(Student("Ada", [92, 85, 78]))
manager.add_student(Student("Mert", [70, 65, 80]))
manager.save()

fresh_manager = GradeManager()   # a brand new instance
fresh_manager.load()
for s in fresh_manager.students:
    print(s.name, s.average())
```
